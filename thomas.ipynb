#### Write a program using lambda functions if the marks is greater than 80 it is distinction if marks is greater than 60, it is first division, if the marks is greater than 40, else it is fail; it is third division my first marks is 90, 

```python
marks = int(input("Enter marks:"))
result=lambda x:"Distinction" if x > 80 and marks<=100 else "First Division" if x > 60 and marks<=80else "Second Division" if x > 40 and marks<=60else "Third Division" if x < 40 and marks>=0else "Fail" if x<40 and x>=0else "invalid input"

print(result(marks))
```
